```python
import pandas as pd
import numpy as np

# Student-Specific Inputs
LAST_NAME = "Solina"
STUDENT_ID = "TUPM-26-2302"
SEED_NUM = 2302  # Derived from the last 4 digits of Student ID
FAVORITE_ARTIST = "BINI"

# Deterministic value generator function using SEED_NUM and character ASCII positions
def generate_value(base_mod, index, char):
    return int((SEED_NUM * (index + 1) + ord(char)) % base_mod)

print("="*60)
print(f"LABORATORY ACTIVITY 4 - DATA STRUCTURES")
print(f"Student: {LAST_NAME.upper()} | ID: {STUDENT_ID} | Seed: {SEED_NUM}")
print("="*60)
```

```python
print("--- EXERCISE 1: ENGINEERING TEST DATA ORGANIZER ---")

# Setup components
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate base data
ex1_records = []
for i, src in enumerate(sources):
    for j, cat in enumerate(categories):
        char = FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)]
        val = generate_value(150, i + j, char) + 10.5
        ex1_records.append({"Source": src, "Category": cat, "Measurement": round(val, 2)})

# Fulfill Data Anomalies Requirements
# 1. Include one repeated test-source/category record
ex1_records.append(ex1_records[0].copy())
# 2. Include multiple measurements under at least two categories
ex1_records.append({"Source": sources[0], "Category": categories[0], "Measurement": 99.4})
ex1_records.append({"Source": sources[1], "Category": categories[1], "Measurement": 124.2})

df1_raw = pd.DataFrame(ex1_records)

# Requirements Processing
distinct_df1 = df1_raw.drop_duplicates()
repeated_df1 = df1_raw[df1_raw.duplicated(keep=False)]
summary_stats = distinct_df1.groupby("Category")["Measurement"].agg(["mean", "min", "max", "count"])
ordered_df1 = distinct_df1.sort_values(by=["Category", "Source"]).reset_index(drop=True)

# Outputs
print("\n[Assessment Data 1] Generated Raw Test Data (First 5 rows):")
print(df1_raw.head().to_string())

print("\n[Assessment Data 3] Repeated Records Detected:")
print(repeated_df1.to_string())

print("\n[Assessment Data 4] Numerical Summaries by Category:")
print(summary_stats.to_string())

print("\n[Final Output] Final Organized and Summarized Test Dataset:")
print(ordered_df1.to_string())
```

```python
print("--- EXERCISE 2: STUDENT PERFORMANCE DATA ANALYZER ---")

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

ex2_records = []
for i, student in enumerate(students):
    for j, subj in enumerate(subjects):
        char = FAVORITE_ARTIST[j % len(FAVORITE_ARTIST)]
        score = 50 + (generate_value(51, i * j, char))
        ex2_records.append({"Student": student, "Subject": subj, "Score": score})

# Fulfill Data Anomalies Requirements
# 1. Include one repeated student-subject record
ex2_records.append(ex2_records[0].copy())
# 2. Include one student record with a missing score
ex2_records.append({"Student": students[2], "Subject": subjects[1], "Score": None})

df2_raw = pd.DataFrame(ex2_records)

# Grading Category Rule
def get_performance_category(score):
    if pd.isna(score): return "Incomplete"
    elif score >= 90:  return "Excellent"
    elif score >= 75:  return "Good"
    elif score >= 60:  return "Pass"
    else:              return "Fail"

df2_raw["Performance Category"] = df2_raw["Score"].apply(get_performance_category)

# Identify repeated and incomplete records
repeated_df2 = df2_raw[df2_raw.duplicated(subset=["Student", "Subject"], keep=False)]
incomplete_df2 = df2_raw[df2_raw["Score"].isna()]

# Summaries & Sorting
subject_summary = df2_raw.dropna().groupby("Subject")["Score"].mean().to_frame(name="Average Score")
ordered_df2 = df2_raw.dropna().sort_values(by=["Student", "Score"], ascending=[True, False]).reset_index(drop=True)

# Outputs
print("\n[Assessment Data 1 & 4] Performance Records with Categories:")
print(df2_raw.head(8).to_string())

print("\n[Assessment Data 3] Repeated / Incomplete Student Records:")
print(f"Incomplete rows count: {len(incomplete_df2)} | Repeated rows count: {len(repeated_df2)}")

print("\n[Final Output] Final Student Performance Report (Sorted):")
print(ordered_df2.to_string())
```

```python
print("--- EXERCISE 3: TRANSACTION HISTORY ANALYZER ---")

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
fixed_prices = {"NOTEBOOK": 45.00, "PEN": 15.00, "USB": 350.00, "HEADPHONES": 899.00, "CALCULATOR": 1200.00}

ex3_records = []
for i, cust in enumerate(customers):
    for j, prod in enumerate(products):
        char = FAVORITE_ARTIST[(i+j) % len(FAVORITE_ARTIST)]
        qty = (generate_value(5, i + j, char) % 5) + 1
        ex3_records.append({"Customer": cust, "Product": prod, "Quantity": qty})

# Fulfill Data Anomalies Requirements
# 1. Include at least one repeated customer-product transaction
ex3_records.append(ex3_records[0].copy())

df3_raw = pd.DataFrame(ex3_records)

# Calculate prices and total transaction values
df3_raw["Price"] = df3_raw["Product"].map(fixed_prices)
df3_raw["Total Value"] = df3_raw["Quantity"] * df3_raw["Price"]

# Processing Summaries
repeated_df3 = df3_raw[df3_raw.duplicated(subset=["Customer", "Product"], keep=False)]
cust_summary = df3_raw.groupby("Customer")["Total Value"].sum().to_frame(name="Total Spent")
prod_summary = df3_raw.groupby("Product")["Quantity"].sum().to_frame(name="Total Units Sold")
overall_total = df3_raw["Total Value"].sum()
ordered_df3 = df3_raw.sort_values(by="Total Value", ascending=False).reset_index(drop=True)

# Outputs
print("\n[Assessment Data 4] Transaction Records Sample:")
print(df3_raw.head(5).to_string())

print("\n[Assessment Data 5 & 6] Summaries:")
print(cust_summary.to_string())
print("\n", prod_summary.to_string())

print(f"\n[Assessment Data 8] Overall Dataset Transaction Summary Value: PHP {overall_total:,.2f}")

print("\n[Final Output] Final Consolidated Transaction Report:")
print(ordered_df3.to_string())
```
